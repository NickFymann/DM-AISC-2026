# Cubic-Dictionary Cancellation of a Tanh PIM Source with Additive White Gaussian Noise

## Scope and reproducibility

This self-contained notebook evaluates the cubic Volterra canceller when the phase-preserving tanh PIM source is observed in complex additive white Gaussian noise (AWGN). The physical AWGN power is set 35 dB below the clean PIM power. Dense ridge-LS and correlation-pruned fixed-noise SBL use the same noisy observation and the same uncorrupted transmit-side references.

The notebook evaluates both observed-signal cancellation and a simulation-only PIM-component diagnostic. All data, noise, estimators, metrics, and figures are generated locally.


```python
from __future__ import annotations

import csv
from dataclasses import dataclass
from itertools import combinations_with_replacement
from pathlib import Path
from time import perf_counter

import matplotlib.pyplot as plt
import numpy as np
from scipy.signal import firwin, lfilter

S = 5 * 2048
FS = 256e6
ORDER = 2048
F_SHIFT = 30e6
F_UP = 90e6
NL_GAIN = 1.2 + 3.7j
GAMMA = 2.0
EPS_NL = 1e-12
SEED = 12345
NOISE_SEED = 54321
NOISE_BELOW_PIM_DB = 35.0

OUT = Path('out') / 'cubic_noisy'
OUT.mkdir(parents=True, exist_ok=True)

plt.rcParams.update({
    'figure.figsize': (11, 4.8),
    'figure.dpi': 110,
    'axes.grid': True,
    'axes.axisbelow': True,
    'grid.alpha': 0.3,
    'axes.spines.top': False,
    'axes.spines.right': False,
})

print(f'S = {S}, Fs = {FS / 1e6:g} MHz, FFT step = {FS / S / 1e3:.2f} kHz')
print(f'Nonlinearity: tanh soft saturation, GAMMA = {GAMMA:g}')
print(f'Observation noise target: AWGN {NOISE_BELOW_PIM_DB:g} dB below clean PIM power')
```

## 1. Tanh source and noisy observation model

The bounded nonlinear source is

$$
g_{\tanh}(s)=G\tanh(\gamma|s|)\frac{s}{|s|+\varepsilon},
$$

and the receiver observation is

$$
y[n]=p[n]+v[n],
$$

where $p[n]$ is the filtered PIM component and $v[n]$ is zero-mean circular complex AWGN. Its variance is selected from the valid clean-PIM power so that $\mathbb{E}|v|^2$ is 35 dB below $\mathbb{E}|p|^2$. The transmit-side reference signals are not corrupted.

Two evaluation views are retained. **Observed cancellation** uses $y=p+v$ in both numerator and residual. **PIM-only cancellation** uses the known simulated $p$ only to diagnose modeling error; it is not available in a measurement-only setting.


```python
@dataclass(frozen=True)
class StandaloneData:
    x_hi_tx: np.ndarray
    x_lo_tx: np.ndarray
    x_hi_at_pim: np.ndarray
    x_lo_at_pim: np.ndarray
    target_clean: np.ndarray
    noise: np.ndarray
    target_filtered: np.ndarray
    noise_variance: float
    lpf: np.ndarray


def nonlinear_transform(sig: np.ndarray) -> np.ndarray:
    mag = np.abs(sig)
    phase = sig / (mag + EPS_NL)
    return NL_GAIN * np.tanh(GAMMA * mag) * phase


def complex_awgn(n: int, variance: float, seed: int) -> np.ndarray:
    rng = np.random.default_rng(seed)
    scale = np.sqrt(variance / 2)
    return scale * (rng.standard_normal(n) + 1j * rng.standard_normal(n))


def generate_standalone_data(seed: int = SEED, noise_seed: int = NOISE_SEED) -> StandaloneData:
    rng = np.random.RandomState(seed)
    s0 = rng.random_sample(S) + 1j * rng.random_sample(S)

    flt = firwin(ORDER + 1, 0.05, window='hamming', scale=True)
    n_fir = np.arange(ORDER + 1)
    flt_hi = flt * np.exp(1j * (2 * np.pi * F_SHIFT * n_fir / FS))
    flt_lo = flt * np.exp(1j * (2 * np.pi * -F_SHIFT * n_fir / FS))

    sig_hi = lfilter(flt_hi, [1.0], s0)
    sig_lo = lfilter(flt_lo, [1.0], s0)
    sig0 = sig_hi + sig_lo

    flt_in = np.array([
        0.186 - 0.023j, -0.098 + 0.117j, 0.291 - 0.338j, -0.607 + 0.628j,
        1.131 - 1.348j, -1.090 + 0.993j, 1.633 - 1.725j, -1.415 + 1.365j,
        0.760 - 0.668j, -0.256 + 0.211j, 0.045 - 0.037j,
    ])

    sig = lfilter(flt_in, [1.0], sig0)
    sig_nl = nonlinear_transform(sig)

    n = np.arange(S)
    shifted = sig_nl * np.exp(1j * (2 * np.pi * F_UP * n / FS))
    lpf = firwin(ORDER + 1, 0.2, window='hamming', scale=True)
    target_clean = lfilter(lpf, [1.0], shifted)
    valid_target_clean = target_clean[2 * ORDER + 16:]
    pim_power = float(np.mean(np.abs(valid_target_clean) ** 2))
    noise_variance = pim_power * 10 ** (-NOISE_BELOW_PIM_DB / 10)
    noise = complex_awgn(target_clean.size, noise_variance, noise_seed)
    target_filtered = target_clean + noise

    rot_hi = np.exp(-1j * (2 * np.pi * F_SHIFT * n / FS))
    rot_lo = np.exp(1j * (2 * np.pi * F_SHIFT * n / FS))
    x_hi_tx = sig_hi * rot_hi
    x_lo_tx = sig_lo * rot_lo

    sig_hi_at_pim = lfilter(flt_in, [1.0], sig_hi)
    sig_lo_at_pim = lfilter(flt_in, [1.0], sig_lo)
    x_hi_at_pim = sig_hi_at_pim * rot_hi
    x_lo_at_pim = sig_lo_at_pim * rot_lo

    return StandaloneData(
        x_hi_tx=x_hi_tx,
        x_lo_tx=x_lo_tx,
        x_hi_at_pim=x_hi_at_pim,
        x_lo_at_pim=x_lo_at_pim,
        target_clean=target_clean,
        noise=noise,
        target_filtered=target_filtered,
        noise_variance=noise_variance,
        lpf=lpf,
    )


data = generate_standalone_data()
noise_mean = np.mean(data.noise)
valid_target_clean = data.target_clean[2 * ORDER + 16:]
valid_noise = data.noise[2 * ORDER + 16:]
pim_power = np.mean(np.abs(valid_target_clean) ** 2)
noise_power = np.mean(np.abs(valid_noise) ** 2)
noise_below_pim_db = 10 * np.log10(pim_power / noise_power)
print(f'target length = {len(data.target_filtered)}, lpf length = {len(data.lpf)}')
print(f'clean PIM E|p|^2 = {pim_power:.4e}')
print(f'target AWGN variance = {data.noise_variance:.4e}')
print(f'added AWGN mean = {noise_mean.real:.3e} + {noise_mean.imag:.3e}j')
print(f'added AWGN empirical E|v|^2 = {noise_power:.4e}; PIM/noise = {noise_below_pim_db:.2f} dB')
```

## 2. Cubic dictionary and dense ridge-LS

The causal cubic candidates are

$$
\phi^{(3)}_{a,b,c}[n]=x_\ell[n-a]x_\ell[n-b]x_h^*[n-c],
\qquad a\le b,
$$

with $a,b,c\in\{0,\ldots,M-1\}$ and $N_3(M)=M^2(M+1)/2$. Every candidate is passed through the receiver low-pass filter. The dense baseline fits the complete dictionary with the scale-adaptive ridge penalty defined in this notebook.

For independent zero-mean noise, the sample cross-term $X_f^H v_f/N_f$ tends to average down as the fitting record grows. The noise itself is not canceled by the PIM model and therefore remains in the observed residual.


```python
def lagged_matrix(x: np.ndarray, lags: list[int]) -> tuple[np.ndarray, np.ndarray]:
    if not lags:
        raise ValueError('lags must not be empty')
    length = x.size
    start = max(0, max(lags))
    stop = min(length, length + min(lags))
    if stop <= start:
        raise ValueError('signal is too short for requested lags')
    n = np.arange(start, stop)
    return np.stack([x[n - lag] for lag in lags], axis=1), n


def im3_basis_lags(
    xa: np.ndarray,
    xb: np.ndarray,
    xc: np.ndarray,
    lags: list[int],
    *,
    symmetric: bool = True,
    selected_cols: np.ndarray | list[int] | None = None,
) -> tuple[np.ndarray, np.ndarray]:
    da, n_a = lagged_matrix(xa, lags)
    db, n_b = lagged_matrix(xb, lags)
    dc, n_c = lagged_matrix(xc, lags)
    if not (np.array_equal(n_a, n_b) and np.array_equal(n_a, n_c)):
        raise RuntimeError('lag grids are inconsistent')

    n_lags = len(lags)
    pairs = (
        list(combinations_with_replacement(range(n_lags), 2))
        if symmetric
        else [(k1, k2) for k1 in range(n_lags) for k2 in range(n_lags)]
    )
    n_cols = len(pairs) * n_lags
    selected = range(n_cols) if selected_cols is None else [int(col) for col in selected_cols]
    cols = []
    for col in selected:
        if col < 0 or col >= n_cols:
            raise ValueError(f'selected column {col} is outside 0..{n_cols - 1}')
        pair_idx, k3 = divmod(col, n_lags)
        k1, k2 = pairs[pair_idx]
        cols.append(da[:, k1] * db[:, k2] * np.conj(dc[:, k3]))
    if not cols:
        raise ValueError('selected_cols must not be empty')
    return np.stack(cols, axis=1), n_a


def filtered_basis(x_mat: np.ndarray, lpf: np.ndarray) -> np.ndarray:
    return lfilter(lpf, [1.0], x_mat, axis=0)


def build_problem(
    data: StandaloneData,
    lags: list[int],
    *,
    use_oracle_refs: bool = False,
    apply_rx_lpf: bool = True,
    discard_samples: int = 2 * ORDER + 16,
    selected_cols: np.ndarray | list[int] | None = None,
) -> tuple[np.ndarray, np.ndarray, np.ndarray]:
    if use_oracle_refs:
        x_lo, x_hi = data.x_lo_at_pim, data.x_hi_at_pim
    else:
        x_lo, x_hi = data.x_lo_tx, data.x_hi_tx

    x_mat, n = im3_basis_lags(x_lo, x_lo, x_hi, lags, symmetric=True, selected_cols=selected_cols)
    if apply_rx_lpf:
        x_mat = filtered_basis(x_mat, data.lpf)

    y = data.target_filtered[n]
    start_by_abs_n = int(np.searchsorted(n, discard_samples))
    start_by_lpf = ORDER if apply_rx_lpf else 0
    start = max(start_by_abs_n, start_by_lpf)
    return x_mat[start:], y[start:], n[start:]


def cancellation_db(target: np.ndarray, residual: np.ndarray) -> float:
    eps = 1e-30
    return 10 * np.log10(np.mean(np.abs(target) ** 2) / (np.mean(np.abs(residual) ** 2) + eps))


def averaged_noise_projection_db(x_mat: np.ndarray, clean: np.ndarray, noise: np.ndarray) -> float:
    eps = 1e-30
    col_norms = np.linalg.norm(x_mat, axis=0)
    keep = col_norms > eps
    x_normed = x_mat[:, keep] / col_norms[keep]
    k = x_normed.shape[0]
    clean_corr = x_normed.conj().T @ clean / k
    noise_corr = x_normed.conj().T @ noise / k
    return 20 * np.log10(np.linalg.norm(noise_corr) / (np.linalg.norm(clean_corr) + eps) + eps)


def solve_wiener_hopf(x_mat: np.ndarray, y: np.ndarray, ridge: float = 1e-4) -> np.ndarray:
    k = x_mat.shape[0]
    r_mat = x_mat.conj().T @ x_mat / k
    r_vec = x_mat.conj().T @ y / k
    reg = ridge * np.trace(r_mat).real / r_mat.shape[0]
    return np.linalg.solve(r_mat + reg * np.eye(r_mat.shape[0]), r_vec)
```

## 3. Fixed-noise SBL and the distinction between two noise concepts

Correlation screening and SBL operate on normalized fitting variables. The likelihood precision $\beta$ is held fixed within each run and controls the Bayesian regression in normalized coordinates. The physically added AWGN is a separate part of the simulated observation. Sweeping $\beta$ therefore does not alter the waveform noise realization; it changes only the inference setting.

The coefficient precisions are updated from the posterior covariance and mean, irrelevant terms are pruned, and the retained coefficients are rescaled before synthesizing the cancellation replica.


```python
@dataclass(frozen=True)
class SblConfig:
    beta: float = 100.0
    eta: float = 0.01
    keep_top: int | None = None
    min_keep: int = 1
    alpha_init: float = 1.0
    alpha_max: float = 1e6
    max_iter: int = 200
    tol: float = 1e-6
    eps: float = 1e-12


@dataclass(frozen=True)
class SblResult:
    h: np.ndarray
    active: np.ndarray
    screened: np.ndarray
    scores: np.ndarray
    alpha: np.ndarray
    mu_norm: np.ndarray
    iterations: int
    converged: bool
    y_norm: float
    col_norms: np.ndarray


def correlation_screen(
    x_normed: np.ndarray,
    y_normed: np.ndarray,
    *,
    eta: float,
    keep_top: int | None,
    min_keep: int,
) -> tuple[np.ndarray, np.ndarray]:
    scores = np.abs(x_normed.conj().T @ y_normed)
    keep = scores > eta

    if keep_top is not None:
        top = np.argsort(scores)[-min(keep_top, scores.size):]
        keep[top] = True

    if keep.sum() < min_keep:
        top = np.argsort(scores)[-min(min_keep, scores.size):]
        keep[:] = False
        keep[top] = True

    return np.flatnonzero(keep), scores


def solve_sbl_pruned(x_mat: np.ndarray, y: np.ndarray, cfg: SblConfig = SblConfig()) -> SblResult:
    if x_mat.ndim != 2:
        raise ValueError('x_mat must be 2-D')
    if y.ndim != 1 or y.size != x_mat.shape[0]:
        raise ValueError('y must be a vector with len(y) == len(X)')

    y_norm = float(np.linalg.norm(y))
    if y_norm <= cfg.eps:
        raise ValueError('target vector is numerically zero')

    col_norms = np.linalg.norm(x_mat, axis=0)
    nonzero_cols = np.flatnonzero(col_norms > cfg.eps)
    if nonzero_cols.size == 0:
        raise ValueError('all dictionary columns are numerically zero')

    x_normed = x_mat[:, nonzero_cols] / col_norms[nonzero_cols]
    y_normed = y / y_norm

    screened_local, scores_local = correlation_screen(
        x_normed,
        y_normed,
        eta=cfg.eta,
        keep_top=cfg.keep_top,
        min_keep=cfg.min_keep,
    )
    screened = nonzero_cols[screened_local]
    scores = np.zeros(x_mat.shape[1], dtype=float)
    scores[nonzero_cols] = scores_local

    x_active = x_normed[:, screened_local]
    active = screened.copy()
    alpha = np.full(x_active.shape[1], cfg.alpha_init, dtype=float)
    prev_mu: np.ndarray | None = None
    converged = False
    mu = np.zeros(x_active.shape[1], dtype=complex)

    for iteration in range(1, cfg.max_iter + 1):
        gram = x_active.conj().T @ x_active
        rhs = x_active.conj().T @ y_normed
        system = cfg.beta * gram + np.diag(alpha)

        sigma = np.linalg.inv(system)
        sigma = (sigma + sigma.conj().T) / 2
        mu = cfg.beta * sigma @ rhs

        gamma_eff = 1.0 - alpha * np.real(np.diag(sigma))
        gamma_eff = np.clip(gamma_eff, cfg.eps, 1.0)
        alpha_new = gamma_eff / (np.abs(mu) ** 2 + cfg.eps)
        alpha_new = np.clip(alpha_new, cfg.eps, 1e16)

        keep = np.isfinite(alpha_new) & (alpha_new <= cfg.alpha_max)
        if keep.sum() == 0:
            keep[np.argmin(alpha_new)] = True

        if keep.sum() < keep.size:
            x_active = x_active[:, keep]
            active = active[keep]
            alpha = alpha_new[keep]
            mu = mu[keep]
            prev_mu = None
            continue

        if prev_mu is not None:
            denom = max(np.linalg.norm(prev_mu), cfg.eps)
            rel = np.linalg.norm(mu - prev_mu) / denom
            if rel < cfg.tol:
                alpha = alpha_new
                converged = True
                break

        prev_mu = mu.copy()
        alpha = alpha_new

    h = np.zeros(x_mat.shape[1], dtype=complex)
    h[active] = (y_norm / col_norms[active]) * mu
    return SblResult(
        h=h,
        active=active,
        screened=screened,
        scores=scores,
        alpha=alpha,
        mu_norm=mu,
        iterations=iteration,
        converged=converged,
        y_norm=y_norm,
        col_norms=col_norms,
    )
```

## 4. Numerical validation of sparse recovery

A small complex sparse-regression problem verifies the SBL solver independently of the noisy PIM experiment. This check isolates numerical inference from signal generation and receiver filtering.


```python
def synthetic_sbl_check() -> None:
    rng = np.random.default_rng(7)
    n_rows, n_cols, n_active = 600, 80, 5
    x = (rng.standard_normal((n_rows, n_cols)) + 1j * rng.standard_normal((n_rows, n_cols))) / np.sqrt(2 * n_rows)
    h_true = np.zeros(n_cols, dtype=complex)
    support = rng.choice(n_cols, n_active, replace=False)
    h_true[support] = rng.standard_normal(n_active) + 1j * rng.standard_normal(n_active)
    y = x @ h_true
    y += 1e-3 * (rng.standard_normal(n_rows) + 1j * rng.standard_normal(n_rows)) / np.sqrt(2)

    res = solve_sbl_pruned(x, y, SblConfig(beta=1e4, eta=0.0, alpha_max=1e8, max_iter=300, tol=1e-7))
    pred = x @ res.h
    print(f'true support size = {n_active}, recovered active = {res.active.size}, iterations = {res.iterations}, converged = {res.converged}')
    print(f'fit quality = {cancellation_db(y, y - pred):.1f} dB')
    print('true support:', sorted(support.tolist()))
    print('top recovered:', sorted(res.active[np.argsort(np.abs(res.h[res.active]))[-n_active:]].tolist()))


synthetic_sbl_check()
```

## 5. Noisy cubic experiment across memory depths

The fitting and test partitions are identical for ridge-LS and SBL. For each memory depth, the table reports cancellation on the observed waveform, PIM-only cancellation, the normalized projection of the added noise onto the dictionary, support statistics, convergence, and timing. The full experiment covers $M\in\{2,4,6,8\}$.


```python
@dataclass(frozen=True)
class ExperimentRow:
    memory: int
    n_cols: int
    k_train: int
    k_test: int
    ls_train_db: float
    ls_test_db: float
    sbl_train_db: float
    sbl_test_db: float
    ls_train_pim_db: float
    ls_test_pim_db: float
    sbl_train_pim_db: float
    sbl_test_pim_db: float
    noise_projection_db: float
    screened: int
    active: int
    iterations: int
    converged: bool
    ls_train_sec: float
    sbl_train_sec: float
    ls_infer_sec: float
    sbl_infer_sec: float


def run_memory_experiment(
    memory: int,
    *,
    sbl_cfg: SblConfig,
    ridge: float = 1e-4,
) -> tuple[ExperimentRow, np.ndarray, np.ndarray, np.ndarray, np.ndarray, np.ndarray, SblResult]:
    lags = list(range(memory))
    x_mat, y, n_valid = build_problem(data, lags, use_oracle_refs=False, apply_rx_lpf=True)
    y_clean = data.target_clean[n_valid]
    noise = data.noise[n_valid]
    k_train = x_mat.shape[0] // 2
    x_train, y_train = x_mat[:k_train], y[:k_train]

    t0 = perf_counter()
    h_ls = solve_wiener_hopf(x_train, y_train, ridge=ridge)
    ls_train_sec = perf_counter() - t0

    t0 = perf_counter()
    sbl_res = solve_sbl_pruned(x_train, y_train, sbl_cfg)
    sbl_train_sec = perf_counter() - t0

    t0 = perf_counter()
    x_inf_ls, y_inf, n_inf = build_problem(data, lags, use_oracle_refs=False, apply_rx_lpf=True)
    pred_ls = x_inf_ls @ h_ls
    ls_infer_sec = perf_counter() - t0

    t0 = perf_counter()
    x_inf_sbl, y_sbl, n_sbl = build_problem(
        data,
        lags,
        use_oracle_refs=False,
        apply_rx_lpf=True,
        selected_cols=sbl_res.active,
    )
    pred_sbl = x_inf_sbl @ sbl_res.h[sbl_res.active]
    sbl_infer_sec = perf_counter() - t0
    if y_inf.shape != y.shape or y_sbl.shape != y.shape:
        raise RuntimeError('inference timing used a different sample window')
    if not (np.array_equal(n_inf, n_valid) and np.array_equal(n_sbl, n_valid)):
        raise RuntimeError('inference timing used a different sample index')

    row = ExperimentRow(
        memory=memory,
        n_cols=x_mat.shape[1],
        k_train=k_train,
        k_test=x_mat.shape[0] - k_train,
        ls_train_db=cancellation_db(y[:k_train], y[:k_train] - pred_ls[:k_train]),
        ls_test_db=cancellation_db(y[k_train:], y[k_train:] - pred_ls[k_train:]),
        sbl_train_db=cancellation_db(y[:k_train], y[:k_train] - pred_sbl[:k_train]),
        sbl_test_db=cancellation_db(y[k_train:], y[k_train:] - pred_sbl[k_train:]),
        ls_train_pim_db=cancellation_db(y_clean[:k_train], y_clean[:k_train] - pred_ls[:k_train]),
        ls_test_pim_db=cancellation_db(y_clean[k_train:], y_clean[k_train:] - pred_ls[k_train:]),
        sbl_train_pim_db=cancellation_db(y_clean[:k_train], y_clean[:k_train] - pred_sbl[:k_train]),
        sbl_test_pim_db=cancellation_db(y_clean[k_train:], y_clean[k_train:] - pred_sbl[k_train:]),
        noise_projection_db=averaged_noise_projection_db(x_train, y_clean[:k_train], noise[:k_train]),
        screened=sbl_res.screened.size,
        active=sbl_res.active.size,
        iterations=sbl_res.iterations,
        converged=sbl_res.converged,
        ls_train_sec=ls_train_sec,
        sbl_train_sec=sbl_train_sec,
        ls_infer_sec=ls_infer_sec,
        sbl_infer_sec=sbl_infer_sec,
    )
    return row, y, y_clean, noise, pred_ls, pred_sbl, sbl_res


def print_table(rows: list[ExperimentRow]) -> None:
    print(
        f"{'M':>3} {'N':>5} {'Ktr':>5} {'LS noisy':>9} {'SBL noisy':>10} "
        f"{'LS PIM':>8} {'SBL PIM':>9} {'XHv/XHp':>9} {'screen':>7} {'active':>7} {'it':>4} "
        f"{'LS fit':>8} {'SBL fit':>8} {'LS inf':>8} {'SBL inf':>8}"
    )
    print('-' * 124)
    for r in rows:
        print(
            f'{r.memory:>3} {r.n_cols:>5} {r.k_train:>5} '
            f'{r.ls_test_db:>9.1f} {r.sbl_test_db:>10.1f} '
            f'{r.ls_test_pim_db:>8.1f} {r.sbl_test_pim_db:>9.1f} '
            f'{r.noise_projection_db:>9.1f} {r.screened:>7} {r.active:>7} {r.iterations:>4} '
            f'{r.ls_train_sec:>8.3f} {r.sbl_train_sec:>8.3f} '
            f'{r.ls_infer_sec:>8.3f} {r.sbl_infer_sec:>8.3f}'
        )


sbl_cfg = SblConfig(beta=100.0, eta=0.01, keep_top=None, min_keep=1, alpha_max=1e6, max_iter=200, tol=1e-6)

rows = []
series = {}
for memory in (2, 4, 6, 8):
    row, y, y_clean, noise, pred_ls, pred_sbl, sbl_res = run_memory_experiment(memory, sbl_cfg=sbl_cfg)
    rows.append(row)
    series[memory] = (y, y_clean, noise, pred_ls, pred_sbl, sbl_res)

print_table(rows)

csv_path = OUT / 'sbl_pruned_tanh_saturation_awgn_pim_minus_35db_results.csv'
with csv_path.open('w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow([
        'nonlinearity', 'gamma', 'noise_below_pim_db', 'noise_variance', 'memory', 'n_cols', 'k_train', 'k_test',
        'ls_train_noisy_db', 'ls_test_noisy_db', 'sbl_train_noisy_db', 'sbl_test_noisy_db',
        'ls_train_pim_db', 'ls_test_pim_db', 'sbl_train_pim_db', 'sbl_test_pim_db',
        'noise_projection_db', 'screened', 'active', 'iterations', 'converged',
        'ls_train_sec', 'sbl_train_sec', 'ls_infer_sec', 'sbl_infer_sec',
    ])
    for r in rows:
        writer.writerow([
            'tanh', GAMMA, NOISE_BELOW_PIM_DB, data.noise_variance, r.memory, r.n_cols, r.k_train, r.k_test,
            r.ls_train_db, r.ls_test_db, r.sbl_train_db, r.sbl_test_db,
            r.ls_train_pim_db, r.ls_test_pim_db, r.sbl_train_pim_db, r.sbl_test_pim_db,
            r.noise_projection_db, r.screened, r.active, r.iterations, r.converged,
            r.ls_train_sec, r.sbl_train_sec, r.ls_infer_sec, r.sbl_infer_sec,
        ])

print(f'CSV saved to {csv_path}')
```

## 6. Spectral decomposition of the noisy residual

The diagnostic spectrum shows the noisy observation, the clean PIM component, the added AWGN, and the estimator residuals. Once the modeled PIM component is suppressed, the observed residual approaches the noise floor. The PIM-only error curve separates model mismatch from the irreducible added noise.


```python
def spectrum_db(x: np.ndarray) -> tuple[np.ndarray, np.ndarray]:
    freqs = np.fft.fftshift(np.fft.fftfreq(x.size, 1.0 / FS)) / 1e6
    mag = 20 * np.log10(np.abs(np.fft.fftshift(np.fft.fft(x))) + 1e-300)
    return freqs, mag


best = max(rows, key=lambda r: r.sbl_test_pim_db)
y_best, y_clean_best, noise_best, pred_ls_best, pred_sbl_best, best_sbl_res = series[best.memory]

freqs, y_noisy_db = spectrum_db(y_best)
_, y_clean_db = spectrum_db(y_clean_best)
_, noise_db = spectrum_db(noise_best)
_, ls_residual_db = spectrum_db(y_best - pred_ls_best)
_, sbl_residual_db = spectrum_db(y_best - pred_sbl_best)
_, sbl_pim_error_db = spectrum_db(y_clean_best - pred_sbl_best)

fig, ax = plt.subplots(figsize=(11, 4.8))
ax.plot(freqs, y_noisy_db, lw=1.0, label='y = clean PIM + AWGN')
ax.plot(freqs, y_clean_db, lw=1.0, label='clean PIM reference')
ax.plot(freqs, noise_db, lw=0.9, ls='--', label='added AWGN only')
ax.plot(freqs, ls_residual_db, lw=1.0, label=f'LS observed residual, M={best.memory}, noisy={best.ls_test_db:.1f} dB')
ax.plot(freqs, sbl_residual_db, lw=1.0, label=f'SBL observed residual, noisy={best.sbl_test_db:.1f} dB')
ax.plot(freqs, sbl_pim_error_db, lw=1.0, ls=':', label=f'SBL PIM-only error, PIM={best.sbl_test_pim_db:.1f} dB')
ax.set_xlim(-28, 28)
ax.set_xlabel('Frequency, MHz')
ax.set_ylabel('20*log10|FFT(.)|, dB')
ax.set_title(f'Tanh saturation + AWGN {NOISE_BELOW_PIM_DB:g} dB below PIM: best SBL M={best.memory}')
ax.legend()
fig.tight_layout()

plot_path = OUT / 'sbl_pruned_tanh_saturation_awgn_pim_minus_35db.png'
fig.savefig(plot_path, dpi=130)
plt.show()
print(f'Best SBL: M={best.memory}, screened={best.screened}/{best.n_cols}, active={best.active}, iterations={best.iterations}')
print(f'Plot saved to {plot_path}')
```

## 7. Independent sweep over fixed likelihood precision

The sweep evaluates $\beta\in\{10^2,10^4,10^6\}$ for $M\in\{2,4,6,8\}$. Model selection considers observed cancellation, PIM-only cancellation, support size, and convergence. The same physical AWGN realization is retained in every run, so differences are attributable to memory and inference settings rather than to a changing noise draw.


```python
beta_rows = []
for beta in (100.0, 1e4, 1e6):
    for memory in (2, 4, 6, 8):
        cfg = SblConfig(beta=beta, eta=0.01, keep_top=None, min_keep=1, alpha_max=1e6, max_iter=500, tol=1e-6)
        row, *_ = run_memory_experiment(memory, sbl_cfg=cfg)
        beta_rows.append((beta, row))

print(
    f"{'beta':>9} {'M':>3} {'LS noisy':>9} {'SBL noisy':>10} {'LS PIM':>8} {'SBL PIM':>9} "
    f"{'XHv/XHp':>9} {'active':>7} {'it':>4} {'conv':>5} "
    f"{'LS fit':>8} {'SBL fit':>8} {'LS inf':>8} {'SBL inf':>8}"
)
print('-' * 122)
for beta, row in beta_rows:
    print(
        f"{beta:>9.0g} {row.memory:>3} {row.ls_test_db:>9.1f} {row.sbl_test_db:>10.1f} "
        f"{row.ls_test_pim_db:>8.1f} {row.sbl_test_pim_db:>9.1f} {row.noise_projection_db:>9.1f} "
        f"{row.active:>7} {row.iterations:>4} {str(row.converged):>5} "
        f"{row.ls_train_sec:>8.3f} {row.sbl_train_sec:>8.3f} "
        f"{row.ls_infer_sec:>8.3f} {row.sbl_infer_sec:>8.3f}"
    )

beta_csv_path = OUT / 'sbl_pruned_tanh_saturation_awgn_pim_minus_35db_beta_sweep.csv'
with beta_csv_path.open('w', newline='', encoding='utf-8') as f:
    writer = csv.writer(f)
    writer.writerow([
        'nonlinearity', 'gamma', 'noise_below_pim_db', 'noise_variance', 'beta', 'memory', 'n_cols',
        'ls_test_noisy_db', 'sbl_test_noisy_db', 'ls_test_pim_db', 'sbl_test_pim_db',
        'noise_projection_db', 'screened', 'active', 'iterations', 'converged',
        'ls_train_sec', 'sbl_train_sec', 'ls_infer_sec', 'sbl_infer_sec',
    ])
    for beta, row in beta_rows:
        writer.writerow([
            'tanh', GAMMA, NOISE_BELOW_PIM_DB, data.noise_variance, beta, row.memory, row.n_cols,
            row.ls_test_db, row.sbl_test_db, row.ls_test_pim_db, row.sbl_test_pim_db,
            row.noise_projection_db, row.screened, row.active, row.iterations, row.converged,
            row.ls_train_sec, row.sbl_train_sec, row.ls_infer_sec, row.sbl_infer_sec,
        ])

fig, ax = plt.subplots(figsize=(8.5, 4.8))
for memory in (2, 4, 6, 8):
    xs = [beta for beta, row in beta_rows if row.memory == memory]
    ys = [row.sbl_test_pim_db for beta, row in beta_rows if row.memory == memory]
    ax.semilogx(xs, ys, 'o-', label=f'SBL PIM-only M={memory}')
ls_m4 = next(row.ls_test_pim_db for beta, row in beta_rows if beta == 100.0 and row.memory == 4)
ax.axhline(ls_m4, color='k', ls=':', lw=1.0, label=f'ridge-LS M=4 PIM-only: {ls_m4:.1f} dB')
ax.set_xlabel('fixed beta')
ax.set_ylabel('PIM-only test suppression, dB')
ax.set_title(f'Tanh saturation + AWGN {NOISE_BELOW_PIM_DB:g} dB below PIM: fixed-noise SBL sensitivity')
ax.legend()
fig.tight_layout()

beta_plot_path = OUT / 'sbl_pruned_tanh_saturation_awgn_pim_minus_35db_beta_sweep.png'
fig.savefig(beta_plot_path, dpi=130)
plt.show()

best_beta, best_row = max(beta_rows, key=lambda item: item[1].sbl_test_pim_db)
print(f'Best sweep result: beta={best_beta:g}, M={best_row.memory}, SBL PIM-only={best_row.sbl_test_pim_db:.1f} dB, noisy={best_row.sbl_test_db:.1f} dB, active={best_row.active}')
print(f'CSV saved to {beta_csv_path}')
print(f'Plot saved to {beta_plot_path}')

best_cfg = SblConfig(beta=best_beta, eta=0.01, keep_top=None, min_keep=1, alpha_max=1e6, max_iter=500, tol=1e-6)
best_row, y_best, y_clean_best, noise_best, pred_ls_best, pred_sbl_best, best_sbl_res = run_memory_experiment(best_row.memory, sbl_cfg=best_cfg)

freqs, y_noisy_db = spectrum_db(y_best)
_, y_clean_db = spectrum_db(y_clean_best)
_, noise_db = spectrum_db(noise_best)
_, ls_residual_db = spectrum_db(y_best - pred_ls_best)
_, sbl_residual_db = spectrum_db(y_best - pred_sbl_best)
_, sbl_pim_error_db = spectrum_db(y_clean_best - pred_sbl_best)

fig, ax = plt.subplots(figsize=(11, 4.8))
ax.plot(freqs, y_noisy_db, lw=1.0, label='y = clean PIM + AWGN')
ax.plot(freqs, y_clean_db, lw=1.0, label='clean PIM reference')
ax.plot(freqs, noise_db, lw=0.9, ls='--', label='added AWGN only')
ax.plot(freqs, ls_residual_db, lw=1.0, label=f'LS observed residual, M={best_row.memory}, noisy={best_row.ls_test_db:.1f} dB')
ax.plot(
    freqs,
    sbl_residual_db,
    lw=1.0,
    label=f'Best SBL observed residual, beta={best_beta:g}, noisy={best_row.sbl_test_db:.1f} dB',
)
ax.plot(freqs, sbl_pim_error_db, lw=1.0, ls=':', label=f'SBL PIM-only error, PIM={best_row.sbl_test_pim_db:.1f} dB')
ax.set_xlim(-28, 28)
ax.set_xlabel('Frequency, MHz')
ax.set_ylabel('20*log10|FFT(.)|, dB')
ax.set_title(f'Tanh saturation + AWGN {NOISE_BELOW_PIM_DB:g} dB below PIM: best fixed-noise SBL from beta sweep')
ax.legend()
fig.tight_layout()

best_plot_path = OUT / 'sbl_pruned_tanh_saturation_awgn_pim_minus_35db_best_beta.png'
fig.savefig(best_plot_path, dpi=130)
plt.show()
print(f'Selected-beta residual plot saved to {best_plot_path}')
```

## Interpretation of the selected operating point

With the deterministic configuration defined above, at $M=4$ and $\beta=10^6$, cubic SBL reaches approximately $24.0$ dB cancellation on the noisy observation and $24.3$ dB on the underlying PIM component. The corresponding ridge-LS values are approximately $22.2$ dB and $22.3$ dB. The small difference between observed and PIM-only cancellation reflects the AWGN that remains after PIM subtraction, while the overall 24 dB ceiling remains governed by nonlinear-order mismatch.
